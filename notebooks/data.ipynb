```python
import pandas as pd

from algotrade.data import fetch_ohlc
```

```python
MARKET_SYMBOL = "btcusd"
STEP = 86400   # 1 day, in seconds
YEARS = 5

OUT_PATH = "../data/tutorial.csv"
```

```python
# Bitstamp caps each request at 1000 candles; fetch_ohlc pages through with
# `start` until it reaches the present.
data = fetch_ohlc(symbol=MARKET_SYMBOL, step=STEP, years=YEARS)

print(f"Fetched {len(data)} daily candles: "
      f"{pd.to_datetime(data['timestamp'].iloc[0], unit='s')} → "
      f"{pd.to_datetime(data['timestamp'].iloc[-1], unit='s')}")
data.head()
```

```python
data.to_csv(OUT_PATH, index=False)
print(f"Saved to {OUT_PATH}")
```
